## 1. Business Understanding

Analisis ini membantu menentukan **ke mana anggaran marketing sebaiknya dialihkan**.

1. Kategori kanal dan campaign mana yang punya ROAS (revenue ÷ biaya) tertinggi, dan mana yang rugi?
2. Seberapa efisien biaya tiap kanal: CPC, CPL, CPA?
3. Di tahap mana funnel paling bocor: impressions → clicks → leads → orders?
4. Bagaimana tren harian dan mingguan biaya dan revenue selama Februari?
5. Apakah perbedaan ROAS antar kanal nyata atau hanya kebetulan?
6. Kalau biaya ditambah, apakah hasilnya masih seefisien rata-ratanya (Marginal ROAS)?
7. Campaign mana yang sebaiknya di-scale, test, maintain, fix, atau cut?
8. Berapa potensi dampak realokasi anggaran, dengan rentang perkiraan yang jujur?

## 2. Import Library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
```

## 3. Import Dataset

Data dibaca dari Google Sheets. Jika gagal atau kolomnya tidak lengkap, otomatis memakai file lokal `Marketing.csv`.

```python
SHEET_ID = "1u1LJ_J9rtWD3rC1iKyAEfpNLv5_mPHPKq0tWgjLQOBA"
KOLOM_WAJIB = {"c_date", "campaign_name", "category", "impressions", "mark_spent",
               "clicks", "leads", "orders", "revenue"}

def muat_data():
    url = f"https://docs.google.com/spreadsheets/d/{SHEET_ID}/export?format=csv"
    try:
        d = pd.read_csv(url)
        assert KOLOM_WAJIB.issubset(d.columns), "Kolom tidak lengkap"
        return d, "Google Sheets"
    except Exception:
        return pd.read_csv("Marketing.csv"), "file lokal (Marketing.csv)"

df_raw, sumber = muat_data()
print("Sumber data:", sumber)
```

## 4. Data Understanding

```python
df = df_raw.copy()

print("Bentuk data:", df.shape)
print(df.dtypes)
print(df.head())
print(df.describe().round(2))

print("Nilai kosong:", df.isna().sum().sum(), "| Duplikat:", df.duplicated().sum())
print("Rentang tanggal:", df["c_date"].min(), "sampai", df["c_date"].max(),
      "|", df["c_date"].nunique(), "hari")
print("Kategori:", sorted(df["category"].unique()))
print("Nama campaign:", sorted(df["campaign_name"].unique()))   # cari ejaan tidak konsisten
print("Baris revenue = 0:", (df["revenue"] == 0).sum(),
      "| leads = 0:", (df["leads"] == 0).sum())
```

**Kesimpulan:** Data berisi 308 baris (11 campaign × 28 hari, 1-28 Februari 2021) tanpa nilai kosong dan duplikat. Tiga hal perlu ditangani: (1) nama campaign `facebOOK_tier2` yang ejaannya tidak konsisten, (2) `c_date` masih bertipe teks, (3) 19 baris revenue = 0 yang perlu ditandai, bukan dihapus.

## 5. Data Preprocessing

Cleaning, feature engineering, dan validasi otomatis. Jika asumsi dilanggar (jumlah baris, total biaya, total revenue), kode berhenti dengan pesan jelas.

```python
def preprocess(df_raw):
    df = df_raw.copy()

    # --- Cleaning ---
    df["campaign_name"] = df["campaign_name"].str.lower().str.strip()
    df["c_date"] = pd.to_datetime(df["c_date"], format="%Y-%m-%d")   # format eksplisit

    # --- Feature engineering ---
    df["week"] = df["c_date"].dt.isocalendar().week.astype(int)
    df["platform"] = df["campaign_name"].str.split("_").str[0]      # facebook, instagram, google, ...
    df["tanpa_revenue"] = df["revenue"] == 0                         # ditandai, TIDAK dihapus

    # --- Validasi otomatis (gagal keras jika data tidak sesuai harapan) ---
    angka = ["impressions", "clicks", "leads", "orders", "mark_spent", "revenue"]
    assert df.shape[0] == 308, f"Jumlah baris tidak sesuai: {df.shape[0]}"
    assert df.isna().sum().sum() == 0, "Ada nilai kosong"
    assert df.duplicated().sum() == 0, "Ada baris duplikat"
    assert df["c_date"].nunique() == 28, "Jumlah hari unik harus 28"
    assert df["campaign_name"].nunique() == 11, "Harus ada 11 campaign setelah dibersihkan"
    assert df["category"].nunique() == 4, "Harus ada 4 kategori"
    assert (df[angka] >= 0).all().all(), "Ada nilai negatif"
    assert abs(df["mark_spent"].sum() - 30_590_879.82) < 1, "Total biaya tidak cocok dengan SQL"
    assert abs(df["revenue"].sum() - 42_889_366.0) < 1, "Total revenue tidak cocok dengan SQL"
    print("Validasi lolos:", df.shape, "| baris tanpa revenue:", int(df["tanpa_revenue"].sum()))
    return df

df = preprocess(df_raw)
```

**Kesimpulan:** Nama campaign distandarkan (huruf kecil), `c_date` diubah ke tanggal dengan format eksplisit, dan ditambah kolom `week`, `platform`, dan `tanpa_revenue`. Baris revenue = 0 tidak dihapus karena menunjukkan campaign yang berbiaya tanpa hasil. Total biaya dan revenue cocok dengan hasil SQL.

## 6. Exploratory Data Analysis (EDA)

### 6.1 Ringkasan per kategori dan campaign

```python
angka = ["impressions", "clicks", "leads", "orders", "mark_spent", "revenue"]

# Per kategori kanal
by_cat = df.groupby("category")[angka].sum()
by_cat["ROAS"] = by_cat["revenue"] / by_cat["mark_spent"]
by_cat["porsi_biaya_%"] = by_cat["mark_spent"] / by_cat["mark_spent"].sum() * 100
by_cat["CPA"] = by_cat["mark_spent"] / by_cat["orders"]
print(by_cat[["mark_spent", "revenue", "ROAS", "porsi_biaya_%", "CPA"]]
      .round(2).sort_values("ROAS", ascending=False))

# Per campaign (tabel `agg` ini dipakai lagi di bagian 7 dan 8)
agg = df.groupby("campaign_name")[["mark_spent", "revenue"]].sum()
agg["ROAS"] = agg["revenue"] / agg["mark_spent"]
agg["spend_share_%"] = agg["mark_spent"] / agg["mark_spent"].sum() * 100
print(agg.round(2).sort_values("ROAS", ascending=False))

# Grafik: ROAS per campaign (garis merah = titik impas ROAS = 1)
ax = agg["ROAS"].sort_values().plot(kind="barh", figsize=(7, 5), title="ROAS per Campaign")
ax.axvline(1, color="red", linestyle="--")
plt.tight_layout(); plt.savefig("roas_campaign.png", dpi=150); plt.show()
```

**Kesimpulan 6.1:** Kategori ber-ROAS tertinggi adalah ___ (___), dan yang di bawah 1 adalah ___ (menghabiskan ___% dari total biaya). Campaign terbaik: ___; terburuk: ___.

#### 6.2 Tren harian dan mingguan

```python
daily = df.groupby("c_date")[["mark_spent", "revenue"]].sum()
daily.plot(figsize=(10, 4), title="Biaya vs Revenue Harian")
plt.tight_layout(); plt.savefig("tren_harian.png", dpi=150); plt.show()

weekly = df.groupby("week")[["mark_spent", "revenue"]].sum()
weekly["ROAS"] = weekly["revenue"] / weekly["mark_spent"]
print(weekly.round(2))
```

**Kesimpulan 6.2:** Biaya memuncak sekitar tanggal ___ lalu ___. ROAS mingguan ___ (stabil / berubah), jadi perbandingan mingguan mencerminkan ___ (jadwal anggaran / performa).

#### 6.3 Funnel per campaign

```python
f = df.groupby("campaign_name")[["impressions", "clicks", "leads", "orders"]].sum()
f["CTR_%"] = f["clicks"] / f["impressions"] * 100
f["click_to_lead_%"] = f["leads"] / f["clicks"] * 100
f["lead_to_order_%"] = f["orders"] / f["leads"] * 100
print(f[["CTR_%", "click_to_lead_%", "lead_to_order_%"]].round(2)
      .sort_values("lead_to_order_%"))
```

**Kesimpulan 6.3:** Tahap funnel yang paling bocor adalah ___. Campaign dengan lead-to-order terendah: ___ (___%); tertinggi: ___ (___%).

#### 6.4 Korelasi

```python
print(df[["mark_spent", "clicks", "leads", "orders", "revenue"]].corr().round(2))
```

**Kesimpulan 6.4:** Biaya berkorelasi ___ dengan revenue (___), dan orders paling erat dengan revenue (___). Korelasi ini menunjukkan hubungan, bukan sebab-akibat.

## 7. Analisis Lanjutan

### 7.1 Marginal ROAS (marginal return on ad spend)

Marginal ROAS = tambahan revenue per tambahan 1 unit biaya, diperkirakan dari kemiringan regresi per campaign. Angka ini adalah **estimasi dari hubungan data**, bukan hasil eksperimen.

```python
# Marginal ROAS = kemiringan regresi revenue terhadap biaya (tambahan revenue per tambahan 1 biaya)
agg["Marginal_ROAS"] = [np.polyfit(g["mark_spent"], g["revenue"], 1)[0]
                        for _, g in df.groupby("campaign_name")]
agg["marginal_vs_rata2"] = agg["Marginal_ROAS"] / agg["ROAS"]
print(agg.round(2).sort_values("ROAS", ascending=False))

# Bukti tambahan: ROAS saat biaya harian rendah, sedang, tinggi (dibagi per campaign)
df["level_biaya"] = df.groupby("campaign_name")["mark_spent"].transform(
    lambda s: pd.qcut(s, 3, labels=["rendah", "sedang", "tinggi"]))
by_level = df.groupby("level_biaya", observed=True)[["revenue", "mark_spent"]].sum()
by_level["ROAS"] = by_level["revenue"] / by_level["mark_spent"]
print(by_level["ROAS"].round(2))
```

**Kesimpulan 7.1:** Campaign `___` punya Marginal ROAS ___ (___× dari rata-ratanya), artinya tambahan biaya cenderung kurang efisien. ROAS di level biaya rendah/sedang/tinggi adalah ___/___/___, yang mengindikasikan ___ (tanda diminishing returns). Catatan: biaya dan revenue sama-sama naik di pertengahan Februari, jadi faktor waktu ikut berpengaruh.

#### 7.2 Apakah perbedaan ROAS antar kanal nyata?

```python
def bootstrap_roas(sub, n=2000, seed=42):
    rng = np.random.default_rng(seed)
    sp = sub["mark_spent"].to_numpy()
    rv = sub["revenue"].to_numpy()
    idx = rng.integers(0, len(sub), size=(n, len(sub)))
    vals = rv[idx].sum(axis=1) / sp[idx].sum(axis=1)
    return np.percentile(vals, [2.5, 97.5])

ci = {cat: bootstrap_roas(g) for cat, g in df.groupby("category")}
ci_df = pd.DataFrame(ci, index=["batas_bawah", "batas_atas"]).T.round(2)
print(ci_df)
```

**Kesimpulan 7.2:** Kanal ___ jelas lebih baik dari yang lain (rentangnya tidak tumpang tindih). Kanal ___ dan ___ masih tumpang tindih, jadi perbedaannya belum meyakinkan. Catatan: bootstrap memperlakukan tiap baris sebagai pengamatan bebas, sehingga rentangnya kemungkinan terlalu sempit.

## 8. Insight & Rekomendasi

### 8.1 Segmentasi campaign (Scale / Test / Maintain / Fix / Cut)

Batas segmentasi adalah pertimbangan analis: ROAS < 0,5 = Cut; 0,5-1,0 = Fix; 1,0-1,5 = Maintain; ≥ 1,5 dengan porsi biaya ≥ 5% = Scale; ≥ 1,5 dengan porsi biaya < 5% = Test.

```python
def beri_label(r):
    if r["ROAS"] < 0.5:
        return "Cut"
    if r["ROAS"] < 1.0:
        return "Fix"
    if r["ROAS"] < 1.5:
        return "Maintain"
    return "Scale" if r["spend_share_%"] >= 5 else "Test"

agg["aksi"] = agg.apply(beri_label, axis=1)
print(agg[["ROAS", "spend_share_%", "aksi"]].round(2).sort_values("ROAS", ascending=False))

fig, ax = plt.subplots(figsize=(8, 5))
warna = {"Scale": "green", "Test": "blue", "Maintain": "gray", "Fix": "orange", "Cut": "red"}
for nama, r in agg.iterrows():
    ax.scatter(r["spend_share_%"], r["ROAS"], s=120, color=warna[r["aksi"]])
    ax.annotate(nama, (r["spend_share_%"], r["ROAS"]), fontsize=8, xytext=(5, 5),
                textcoords="offset points")
ax.axhline(1, color="red", linestyle="--", linewidth=1)
ax.set_xlabel("Porsi biaya (%)"); ax.set_ylabel("ROAS")
ax.set_title("Segmentasi campaign: ROAS vs porsi biaya")
plt.tight_layout(); plt.savefig("segmentasi_campaign.png", dpi=150); plt.show()
```

**Kesimpulan 8.1:** Scale: ___. Test: ___. Maintain: ___. Fix: ___. Cut: ___. Campaign berlabel Fix dan Cut saat ini menghabiskan ___% dari total biaya.

### 8.2 Skenario realokasi anggaran

Biaya digeser dari campaign ber-ROAS di bawah 1 ke tiga campaign ber-ROAS tertinggi. **Konservatif** memakai Marginal ROAS, **optimistis** memakai ROAS rata-rata. Hasilnya berupa rentang, bukan satu angka.

```python
rendah = agg[agg["ROAS"] < 1].index
top3 = agg["ROAS"].nlargest(3).index
bobot = agg.loc[top3, "mark_spent"] / agg.loc[top3, "mark_spent"].sum()
total_rev = agg["revenue"].sum()

def skenario(persen, kolom):
    geser = agg.loc[rendah, "mark_spent"].sum() * persen
    hilang = (agg.loc[rendah, "mark_spent"] * persen * agg.loc[rendah, kolom]).sum()
    dapat = (geser * bobot * agg.loc[top3, kolom]).sum()
    return (dapat - hilang) / total_rev * 100

hasil = pd.DataFrame({
    "geser_biaya_%": [10, 20, 30],
    "konservatif_%": [skenario(p / 100, "Marginal_ROAS") for p in (10, 20, 30)],
    "optimistis_%":  [skenario(p / 100, "ROAS") for p in (10, 20, 30)],
}).round(1)
print(hasil)
```

**Kesimpulan 8.2:** Menggeser 20% biaya dari campaign ber-ROAS < 1 ke tiga campaign terbaik diperkirakan menambah ___% sampai ___% dari total revenue. Ini **perkiraan batas atas**: simulasi mengasumsikan hubungan biaya-revenue stabil, padahal hasil nyata bisa lebih rendah.

### 8.3 Tabel ringkasan keputusan

```python
ringkasan = agg.join(f[["CTR_%", "click_to_lead_%", "lead_to_order_%"]]).round(2)
print(ringkasan[["ROAS", "Marginal_ROAS", "spend_share_%", "lead_to_order_%", "aksi"]]
      .sort_values("ROAS", ascending=False))
```

**Rekomendasi:**
1. **Scale bertahap** campaign berlabel Scale (___) sambil memantau Marginal ROAS-nya.
2. **Hentikan** campaign berlabel Cut (___) dan **perbaiki atau kurangi** campaign berlabel Fix (___).
3. **Uji coba** menaikkan anggaran kecil di campaign berlabel Test (___).
4. **Perbaiki tahap funnel yang bocor** (___) di campaign `___`.
5. **Geser anggaran bertahap**, mulai 10%, lalu ukur ulang sebelum menggeser lebih besar.
